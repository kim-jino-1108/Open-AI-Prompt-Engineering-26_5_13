# 4. Articulate the desired output format through examples

```python
from openai import OpenAI

client = OpenAI()

# ============================================================
# 4. Show and tell
#    Models respond better when shown specific format
#    requirements. This also makes it easier to
#    programmatically parse out multiple outputs reliably.
# ============================================================

text = """
In a recent interview, Sam Altman, CEO of OpenAI, discussed the growing competition
with Microsoft and Google in the AI space. He highlighted the breakthroughs achieved
with DALL-E and ChatGPT, emphasizing that AGI and AI safety remain the most critical
challenges for the industry going forward.
"""


# --------------------------------------------------
# ❌ Less effective: no output format specified
# --------------------------------------------------

response_bad = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "Extract the entities mentioned in the text below. "
                "Extract the following 4 entity types: "
                "company names, people names, specific topics and themes.\n\n"
                f"Text: {text}"
            ),
        },
    ],
)

print("=== Less Effective ===")
print(response_bad.choices[0].message.content)


# --------------------------------------------------
# ✅ Better: show the model the exact desired format
# --------------------------------------------------

response_good = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "Extract the important entities mentioned in the text below. "
                "First extract all company names, then extract all people names, "
                "then extract specific topics which fit the content and finally "
                "extract general overarching themes.\n\n"
                "Desired format:\n"
                "Company names: <comma_separated_list_of_company_names>\n"
                "People names: -||-\n"
                "Specific topics: -||-\n"
                "General themes: -||-\n\n"
                f"Text: {text}"
            ),
        },
    ],
    temperature=0.1,
    max_tokens=300,
    top_p=1,
    frequency_penalty=0,
)

print("\n=== Better ===")
print(response_good.choices[0].message.content)
```

```python
from openai import OpenAI

client = OpenAI()

# ============================================================
# 4. Show and tell
#    모델에게 원하는 출력 형식을 직접 보여주면
#    더 정확하고 파싱하기 쉬운 결과를 얻을 수 있습니다.
# ============================================================

text = """
OpenAI의 샘 알트만 CEO는 최근 마이크로소프트, 구글과의 AI 경쟁에 대해 언급하며
DALL-E와 ChatGPT의 발전 방향을 설명했습니다.
그는 AGI(일반 인공지능)와 AI 안전성이 핵심 과제라고 강조했습니다.
"""


# --------------------------------------------------
# ❌ 덜 효과적: 출력 형식을 지정하지 않은 프롬프트
# --------------------------------------------------

response_bad = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "아래 텍스트에서 언급된 개체를 추출해줘. "
                "회사명, 인물명, 특정 주제, 테마 4가지 유형을 추출해줘.\n\n"
                f"Text: {text}"
            ),
        },
    ],
)

print("=== 덜 효과적 ===")
print(response_bad.choices[0].message.content)


# --------------------------------------------------
# ✅ 더 나은 방법: 원하는 출력 형식을 명시적으로 지정
# --------------------------------------------------

response_good = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "아래 텍스트에서 중요한 개체를 추출해줘. "
                "먼저 모든 회사명을 추출하고, 그 다음 인물명을 추출하고, "
                "내용에 맞는 특정 주제를 추출하고, "
                "마지막으로 전반적인 테마를 추출해줘.\n\n"
                "원하는 형식:\n"
                "회사명: <쉼표로_구분된_회사명_목록>\n"
                "인물명: -||-\n"
                "특정 주제: -||-\n"
                "전반적 테마: -||-\n\n"
                f"Text: {text}"
            ),
        },
    ],
    temperature=0.1,
    max_tokens=300,
    top_p=1,
    frequency_penalty=0,
)

print("\n=== 더 나은 방법 ===")
print(response_good.choices[0].message.content)
```

# 5. Start with zero-shot, then few-shot, neither of them worked, then fine-tune

```python
from openai import OpenAI

client = OpenAI()

# ============================================================
# 5. Few-shot prompting
#    Provide a couple of examples to guide the model.
#    Few-shot examples help the model understand the exact
#    output format and style you expect.
# ============================================================

text = (
    "NASA's James Webb Space Telescope has captured unprecedented images of distant "
    "galaxies, enabling astronomers to study the early universe in remarkable detail. "
    "The telescope uses infrared technology to peer through cosmic dust and observe "
    "star formation processes billions of light-years away."
)


# --------------------------------------------------
# ✅ Zero-shot: no examples provided
# --------------------------------------------------

response_zero_shot = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "Extract keywords from the below text.\n\n"
                f"Text: {text}\n\n"
                "Keywords:"
            ),
        },
    ],
    temperature=0.1,
    max_tokens=100,
    top_p=1,
    frequency_penalty=0,
)

print("=== Zero-shot ===")
print(response_zero_shot.choices[0].message.content)


# --------------------------------------------------
# ✅ Few-shot: provide a couple of examples
# --------------------------------------------------

response_few_shot = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "Extract keywords from the corresponding texts below.\n\n"
                "Text 1: Stripe provides APIs that web developers can use to integrate "
                "payment processing into their websites and mobile applications.\n"
                "Keywords 1: Stripe, payment processing, APIs, web developers, "
                "websites, mobile applications\n"
                "##\n"
                "Text 2: OpenAI has trained cutting-edge language models that are very "
                "good at understanding and generating text. Our API provides access to "
                "these models and can be used to solve virtually any task that involves "
                "processing language.\n"
                "Keywords 2: OpenAI, language models, text processing, API.\n"
                "##\n"
                f"Text 3: {text}\n"
                "Keywords 3:"
            ),
        },
    ],
    temperature=0.1,
    max_tokens=100,
    top_p=1,
    frequency_penalty=0,
)

print("\n=== Few-shot ===")
print(response_few_shot.choices[0].message.content)
```

```python
from openai import OpenAI

client = OpenAI()

# ============================================================
# 5. Few-shot 프롬프팅
#    몇 가지 예시를 제공하여 모델을 올바른 방향으로 유도합니다.
#    Few-shot 예시는 모델이 원하는 출력 형식과 스타일을
#    정확히 이해하는 데 도움을 줍니다.
# ============================================================

text = (
    "NASA의 제임스 웹 우주망원경은 먼 은하의 전례 없는 이미지를 포착하여 "
    "천문학자들이 초기 우주를 놀라운 수준으로 연구할 수 있게 해주었습니다. "
    "이 망원경은 적외선 기술을 사용해 우주 먼지를 통과하여 수십억 광년 떨어진 "
    "별 형성 과정을 관측합니다."
)


# --------------------------------------------------
# ✅ Zero-shot: 예시 없이 바로 지시
# --------------------------------------------------

response_zero_shot = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "아래 텍스트에서 키워드를 추출해줘.\n\n"
                f"텍스트: {text}\n\n"
                "키워드:"
            ),
        },
    ],
    temperature=0.1,
    max_tokens=100,
    top_p=1,
    frequency_penalty=0,
)

print("=== Zero-shot ===")
print(response_zero_shot.choices[0].message.content)


# --------------------------------------------------
# ✅ Few-shot: 예시 2개를 제공하여 형식 유도
# --------------------------------------------------

response_few_shot = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "아래 각 텍스트에서 키워드를 추출해줘.\n\n"
                "텍스트 1: 스트라이프는 웹 개발자들이 웹사이트와 모바일 애플리케이션에 "
                "결제 처리를 통합할 수 있도록 API를 제공합니다.\n"
                "키워드 1: 스트라이프, 결제 처리, API, 웹 개발자, 웹사이트, 모바일 애플리케이션\n"
                "##\n"
                "텍스트 2: OpenAI는 텍스트 이해 및 생성에 매우 뛰어난 최첨단 언어 모델을 "
                "훈련시켰습니다. 저희 API는 이러한 모델에 대한 접근을 제공하며 언어 처리와 "
                "관련된 거의 모든 작업을 해결하는 데 사용할 수 있습니다.\n"
                "키워드 2: OpenAI, 언어 모델, 텍스트 처리, API\n"
                "##\n"
                f"텍스트 3: {text}\n"
                "키워드 3:"
            ),
        },
    ],
    temperature=0.1,
    max_tokens=100,
    top_p=1,
    frequency_penalty=0,
)

print("\n=== Few-shot ===")
print(response_few_shot.choices[0].message.content)
```

# 6. Reduce “fluffy” and imprecise descriptions

```python
from openai import OpenAI

client = OpenAI()

# ============================================================
# 6. Specify length precisely
#    Instead of vague length descriptors, give the model
#    an exact format: number of sentences, words, or paragraphs.
# ============================================================

product = (
    "AeroDesk Pro - an electric height-adjustable standing desk with "
    "memory presets, built-in USB-C charging ports, and a whisper-quiet motor. "
    "Made from sustainable bamboo with a weight capacity of 350 lbs."
)


# --------------------------------------------------
# ❌ Less effective: vague length instruction
# --------------------------------------------------

response_bad = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "The description for this product should be fairly short, "
                "a few sentences only, and not too much more.\n\n"
                f"Product: {product}"
            ),
        },
    ],
    temperature=0.7,
    max_tokens=300,
    top_p=1,
    frequency_penalty=0,
)

print("=== Less Effective ===")
print(response_bad.choices[0].message.content)


# --------------------------------------------------
# ✅ Better: specify exact length format
# --------------------------------------------------

response_good = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "Use a 3 to 5 sentence paragraph to describe this product.\n\n"
                f"Product: {product}"
            ),
        },
    ],
    temperature=0.7,
    max_tokens=300,
    top_p=1,
    frequency_penalty=0,
)

print("\n=== Better ===")
print(response_good.choices[0].message.content)
```

# 7. Instead of just saying what not to do, say what to do instead

```python
from openai import OpenAI

client = OpenAI()

# ============================================================
# 7. Prefer positive instructions over prohibitions
#    Instead of telling the model what NOT to do,
#    tell it what TO do instead. This leads to more
#    helpful and specific responses.
# ============================================================

customer_message = "I can't log in to my account."


# --------------------------------------------------
# ❌ Less effective: using prohibitions (DO NOT)
# --------------------------------------------------

response_bad = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": (
                "The following is a conversation between an Agent and a Customer. "
                "DO NOT ASK USERNAME OR PASSWORD. DO NOT REPEAT."
            ),
        },
        {
            "role": "user",
            "content": f"Customer: {customer_message}\nAgent:",
        },
    ],
    temperature=0.7,
    max_tokens=300,
    top_p=1,
    frequency_penalty=0,
)

print("=== Less Effective ===")
print(response_bad.choices[0].message.content)


# --------------------------------------------------
# ✅ Better: positive instructions with clear alternative
# --------------------------------------------------

response_good = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": (
                "The following is a conversation between an Agent and a Customer. "
                "The agent will attempt to diagnose the problem and suggest a solution, "
                "whilst refraining from asking any questions related to PII. "
                "Instead of asking for PII, such as username or password, "
                "refer the user to the help article www.samplewebsite.com/help/faq"
            ),
        },
        {
            "role": "user",
            "content": f"Customer: {customer_message}\nAgent:",
        },
    ],
    temperature=0.7,
    max_tokens=300,
    top_p=1,
    frequency_penalty=0,
)

print("\n=== Better ===")
print(response_good.choices[0].message.content)
```

# 8. Code Generation Specific - Use “leading words” to nudge the model toward a particular pattern

```python
from openai import OpenAI

client = OpenAI()

# ============================================================
# 8. Use leading words to guide the model's output
#    Adding "import" hints to the model that it should start
#    writing in Python. (Similarly "SELECT" is a good hint
#    for the start of a SQL statement.)
# ============================================================


# --------------------------------------------------
# ❌ Less effective: no leading word to guide output
# --------------------------------------------------

response_bad = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "# Write a simple python function that\n"
                "# 1. Ask me for a number in mile\n"
                "# 2. It converts miles to kilometers"
            ),
        },
    ],
    temperature=0,
    max_tokens=300,
    top_p=1,
    frequency_penalty=0,
)

print("=== Less Effective ===")
print(response_bad.choices[0].message.content)


# --------------------------------------------------
# ✅ Better: add "import" as a leading word to signal Python
# --------------------------------------------------

response_good = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "# Write a simple python function that\n"
                "# 1. Ask me for a number in mile\n"
                "# 2. It converts miles to kilometers\n\n"
                "import"
            ),
        },
    ],
    temperature=0,
    max_tokens=300,
    top_p=1,
    frequency_penalty=0,
)

print("\n=== Better ===")
print("import" + response_good.choices[0].message.content)
```

```python
from openai import OpenAI

client = OpenAI()

# ============================================================
# 8. 리딩 워드로 모델의 출력 방향 유도하기
#    "import"를 추가하면 모델이 Python으로 작성해야 한다는
#    힌트를 줍니다. (SQL이라면 "SELECT"가 같은 역할을 합니다.)
# ============================================================


# --------------------------------------------------
# ❌ 덜 효과적: 리딩 워드 없이 지시만 작성
# --------------------------------------------------

response_bad = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "# 간단한 파이썬 함수를 작성해줘\n"
                "# 1. 마일 단위 숫자를 입력받아\n"
                "# 2. 킬로미터로 변환해줘"
            ),
        },
    ],
    temperature=0,
    max_tokens=300,
    top_p=1,
    frequency_penalty=0,
)

print("=== 덜 효과적 ===")
print(response_bad.choices[0].message.content)


# --------------------------------------------------
# ✅ 더 나은 방법: "import"로 Python 코드 시작을 유도
# --------------------------------------------------

response_good = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": (
                "# 간단한 파이썬 함수를 작성해줘\n"
                "# 1. 마일 단위 숫자를 입력받아\n"
                "# 2. 킬로미터로 변환해줘\n\n"
                "import"
            ),
        },
    ],
    temperature=0,
    max_tokens=300,
    top_p=1,
    frequency_penalty=0,
)

print("\n=== 더 나은 방법 ===")
print("import" + response_good.choices[0].message.content)
```

# 9. Use the Generate Anything feature

```python
from openai import OpenAI

# OpenAI client 생성
client = OpenAI()

# Generate Anything 스타일 입력
task_description = """
고객 문의 이메일에 답변하는 AI 어시스턴트용 프롬프트를 만들어줘.

조건:
- 항상 공손한 존댓말 사용
- 고객 문의 내용을 먼저 요약
- 해결 방법을 단계별로 안내
- 모르는 내용은 추측하지 말기
- 마지막에는 추가 문의 가능하다고 안내
"""

# Chat Completions API 호출
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": "당신은 Prompt Engineering 전문가입니다."
        },
        {
            "role": "user",
            "content": f"""
다음 작업 설명을 기반으로
실제 사용할 수 있는 고품질 프롬프트를 생성해줘.

작업 설명:
{task_description}
"""
        }
    ],
    temperature=0.7,
    max_completion_tokens=500
)

# 생성된 프롬프트 출력
print("=== 생성된 프롬프트 ===")
print(response.choices[0].message.content)
```
